### 01. Chargement les modèles existants

```python
import joblib
import os

# chemins des modèles d’origine
models_paths = {
    "rf_dpe_sans_conso": r"C:\Users\km_sa\DEV\m2_enedis_dpe_app\models\DPE-Sans Conso\rf_dpe_predictif.joblib",
    "rf_dpe_avec_conso": r"C:\Users\km_sa\DEV\m2_enedis_dpe_app\models\DPE-Avec Conso\rf_dpe_conso_predictif.joblib",
    "rf_conso_final": r"C:\Users\km_sa\DEV\m2_enedis_dpe_app\models\Conso\rf_conso_final.joblib"
}

# dossier de sortie
output_dir = r"C:\Users\km_sa\DEV\m2_enedis_dpe_app\models\Compressed"
os.makedirs(output_dir, exist_ok=True)

# chargement des modèles
models = {name: joblib.load(path) for name, path in models_paths.items()}

print("Tous les modèles ont été chargés avec succès.")
```

```python
for name, model in models.items():
    compressed_path = os.path.join(output_dir, f"{name}_compressed.joblib")
    joblib.dump(model, compressed_path, compress=("xz", 6))
    size_mb = os.path.getsize(compressed_path) / (1024 * 1024)
    print(f"{name} compressé → {size_mb:.1f} Mo")
```

```python
import pandas as pd

sizes = []
for name, path in models_paths.items():
    original_size = os.path.getsize(path) / (1024 * 1024)
    compressed_size = os.path.getsize(os.path.join(output_dir, f"{name}_compressed.joblib")) / (1024 * 1024)
    sizes.append({"Modèle": name, "Original (Mo)": round(original_size, 1), "Compressé (Mo)": round(compressed_size, 1)})

df_sizes = pd.DataFrame(sizes)
display(df_sizes)
```
