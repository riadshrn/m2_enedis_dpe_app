#### nettoyage

```python
#garder les colonnes avec mmoins de 95% de valeurs manquantes
valeur_ratio = df_existant.isna().mean() * 100
df_existant = df_existant.loc[:, valeur_ratio < 95]  # garde les colonnes avec <95% de NaN
```

```python
#remplacer les valeurs manquantes par la moyenne de la colonne
for col in df_existant.select_dtypes(include=['float64', 'int64']).columns:
    df_existant[col].fillna(df_existant[col].mean(), inplace=True)

#remplacer les valeurs manquantes par la valeur la plus fréquente de la colonne
for col in df_existant.select_dtypes(include=['object']).columns:
    df_existant[col].fillna(df_existant[col].mode()[0], inplace=True)    
```
