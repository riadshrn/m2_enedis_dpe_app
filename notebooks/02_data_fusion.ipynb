```python
import pandas as pd

# === Chargement des fichiers ===
df_existant = pd.read_csv("../data/df_adem_69.csv", dtype=str)
df_neuf = pd.read_csv("../data/df_adem_neuf_69.csv", dtype=str)
```

### Ajout d'une colonne indiquant la source

```python
df_existant["type_logement_source"] = "existant"
df_neuf["type_logement_source"] = "neuf"
```

### Harmonisation des colonnes avant fusion

```python
colonnes_communes = list(set(df_existant.columns) & set(df_neuf.columns))
print(f"{len(colonnes_communes)} colonnes communes détectées")
```

### Fusion

```python
df_all = pd.concat([df_existant[colonnes_communes], df_neuf[colonnes_communes]], ignore_index=True)
print(f"Fusion réussie : {len(df_all)} lignes, {len(df_all.columns)} colonnes")
```

```python
df_all.groupby("type_logement_source")["annee_construction"].describe()
```

### Export final

```python
df_all.to_csv("../data/df_adem_merge_69.csv", index=False, encoding="utf-8")
print("Fichier fusionné enregistré : ../data/df_adem_merge_69.csv")
```
